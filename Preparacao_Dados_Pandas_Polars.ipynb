## 🧹 7. Problema de formatação — IDs

Vamos verificar os identificadores.

Observe se aparecem valores como:

```text
"CNR5884300"
```

As aspas podem fazer parte do conteúdo e podem ser removidas.


````python
df_pd[["Booking ID", "Customer ID"]].head(10)
````

````python
# Pandas
df_pd["Booking ID"] = df_pd["Booking ID"].str.replace('"', '', regex=False)
df_pd["Customer ID"] = df_pd["Customer ID"].str.replace('"', '', regex=False)

df_pd[["Booking ID", "Customer ID"]].head()
````

````python
# Polars
df_pl = df_pl.with_columns(
    pl.col("Booking ID").str.strip_chars('"'),
    pl.col("Customer ID").str.strip_chars('"')
)

df_pl.select(["Booking ID", "Customer ID"]).head()
````

````python
# 1. Imprimir diretamente a primeira string (não renderiza como HTML do DataFrame)
print("Valor real na memória:", df_pl["Booking ID"][0])
# Saída: Valor real na memória: CNR5884300
````

````python
###Pandas: Mostra textos sem aspas na tabela renderizada.

###Polars: Coloca aspas em dados str na tabela do notebook para explicitar que aquele valor é uma string e não um número ou identificador sem tipo.
````

## 📅 8. Convertendo `Date`

A coluna `Date` deve ser tratada como data.


````python
# Pandas
df_pd["Date"] = pd.to_datetime(df_pd["Date"], errors="coerce") #coerce para lidar com valores inválidos, 
                                                                #caso seja necessário por exemplo, se houver valores nulos ou em formatos diferentes do esperado ele não vai gerar erro, mas sim retornar nulo para esses casos

df_pd[["Date"]].head()
````

````python
# Polars
df_pl = df_pl.with_columns(   #with_columns para criar novas colunas ou modificar colunas existentes, 
                                #ele percorre todas as colunas especificadas e aplica as transformações desejadas, retornando um novo DataFrame com as alterações
    pl.col("Date").str.to_date(strict=False)#strict=False para lidar com valores inválidos, 
                                                                #caso seja necessário por exemplo, se houver valores nulos ou em formatos diferentes do esperado ele não vai gerar erro, mas sim retornar nulo para esses casos

)

df_pl.select("Date").head()
````

## ⏰ 9. Trabalhando com `Time`

O horário pode ser convertido para um tipo temporal.

Para Feature Engineering, também podemos posteriormente extrair diretamente a hora.


````python
# Pandas
df_pd["Time"] = pd.to_datetime(
    df_pd["Time"],
    format="%H:%M:%S",
    errors="coerce"
)

df_pd["Time"].head()
````

````python
# Polars
df_pl = df_pl.with_columns(
    pl.col("Time").str.to_time(format="%H:%M:%S", strict=False) )

df_pl.select("Time").head()
````

# 🧩 18. Feature Engineering

Depois de entender os dados, podemos criar novas variáveis.

Vamos criar:

- `Ano`
- `Mes`
- `Dia`
- `Dia_Semana`
- `Hora`
- `Categoria_Distancia`


````python
# Pandas — calendário
df_pd["Ano"] = df_pd["Date"].dt.year #dt. extractor de data e hora do pandas, permite acessar atributos como ano, mês, dia, hora, minuto, segundo, etc.
df_pd["Mes"] = df_pd["Date"].dt.month
df_pd["Dia"] = df_pd["Date"].dt.day
df_pd["Dia_Semana"] = df_pd["Date"].dt.dayofweek

# Pandas — hora
df_pd["Hora"] = df_pd["Time"].dt.hour

df_pd[[
    "Date", "Time", "Ano", "Mes", "Dia", "Dia_Semana", "Hora"
]].head()
````

````python
# Polars — calendário e hora
df_pl = df_pl.with_columns(
    pl.col("Date").dt.year().alias("Ano"),
    pl.col("Date").dt.month().alias("Mes"),
    pl.col("Date").dt.day().alias("Dia"),
    pl.col("Date").dt.weekday().alias("Dia_Semana"),
    pl.col("Time").dt.hour().alias("Hora")
)

df_pl.select([
    "Date", "Time", "Ano", "Mes", "Dia", "Dia_Semana", "Hora"
]).head()
````

## 📏 19. Criando categorias de distância

````python
# Pandas
df_pd["Categoria_Distancia"] = pd.cut(
    df_pd["Ride Distance"],
    bins=[-float("inf"), 5, 15, float("inf")],
    labels=["Curta", "Média", "Longa"]
)

df_pd[["Ride Distance", "Categoria_Distancia"]].head()
````

````python
# Polars
df_pl = df_pl.with_columns(
    # strict=False converte textos invalidos como "null" para o tipo null nativo do Polars
    pl.col("Ride Distance").cast(pl.Float64, strict=False).alias("Ride Distance")
).with_columns(
    pl.when(pl.col("Ride Distance") <= 5)
      .then(pl.lit("Curta"))
      .when(pl.col("Ride Distance") <= 15)
      .then(pl.lit("Média"))
      .when(pl.col("Ride Distance") > 15)
      .then(pl.lit("Longa"))
      .otherwise(pl.lit("Sem Informação")) # Trata os valores que eram nulos/invalidos
      .alias("Categoria_Distancia")
)

df_pl.select([
    "Ride Distance", "Categoria_Distancia"
]).head(10)
````

## 📊 22. Agregações — valor médio por tipo de veículo

# 🧠 23. Comparação Pandas × Polars

| Operação | Pandas | Polars |
|---|---|---|
| Ler CSV | `pd.read_csv()` | `pl.read_csv()` |
| Primeiras linhas | `.head()` | `.head()` |
| Dimensões | `.shape` | `.shape` |
| Colunas | `.columns` | `.columns` |
| Tipos | `.dtypes` | `.schema` |
| Nulos | `.isna()` | `.null_count()` |
| Filtrar | `df[condição]` | `.filter()` |
| Agrupar | `.groupby()` | `.group_by()` |
| Ordenar | `.sort_values()` | `.sort()` |
| Criar coluna | `df["nova"] = ...` | `.with_columns()` |
| Estatísticas | `.describe()` | `.describe()` |
| Correlação | `.corr()` | `pl.corr()` |

> O objetivo não é escolher uma biblioteca "vencedora", mas compreender diferentes formas de trabalhar com dados tabulares.
