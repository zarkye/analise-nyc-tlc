```python
import pandas as pd
import numpy as np

df = pd.read_csv('uber.csv')

colunas_id = ['Booking ID', 'Customer ID']
for col in colunas_id:
    df[col] = df[col].astype(str).str.replace('"', '')

df['Data_Hora'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], errors='coerce')
df = df.drop(columns=['Date', 'Time'])

colunas_numericas = [
    'Avg VTAT', 'Avg CTAT', 'Booking Value', 
    'Ride Distance', 'Driver Ratings', 'Customer Rating'
]

for col in colunas_numericas:
    df[col] = pd.to_numeric(df[col], errors='coerce')

display(df.head())
```

```python
status_geral = df['Booking Status'].value_counts(normalize=True) * 100
print(status_geral.round(2).astype(str) + "%")
print("\n")

print("TAXA DE 'Driver not Found' POR VEICULO")
df_sem_motorista = df[df['Booking Status'] == 'No Driver Found']
sem_motorista_por_veiculo = df_sem_motorista['Vehicle Type'].value_counts()
print(sem_motorista_por_veiculo)
print("\n")

df_sucesso = df[df['Booking Status'] == 'Completed']

print("MÉTRICAS DAS CORRIDAS COMPLETADAS")
resumo_financeiro = df_sucesso.groupby('Vehicle Type').agg(
    Faturamento_Total=('Booking Value', 'sum'),
    Ticket_Medio=('Booking Value', 'mean'),
    Distancia_Media_km=('Ride Distance', 'mean'),
    Tempo_Espera_Medio_Min=('Avg VTAT', 'mean')
).round(2)

display(resumo_financeiro)
```
